```python
import requests
import time

# 1. API Endpoint URL (Where the live iss-now.json data comes from)
url = "http://api.open-notify.org/iss-now.json"

print("--- ISS Live Tracker Has Started ---")

# Running a loop to fetch the live location multiple times
for i in range(3):  # This will check the location 3 times
    response = requests.get(url)

    # Check if the connection to the API was successful (Status Code 200)
    if response.status_code == 200:
        data = response.json()

        # Extracting the required values from the JSON data
        latitude = data["iss_position"]["latitude"]
        longitude = data["iss_position"]["longitude"]
        timestamp = data["timestamp"]

        # Displaying the clean, formatted data on the screen
        print(f"\n[Check {i+1}] ISS Live Location:")
        print(f"-> Latitude: {latitude}")
        print(f"-> Longitude: {longitude}")

    else:
        print("Error: Unable to connect to the API or internet issue!")

    time.sleep(5)  # Wait for 5 seconds before checking again
```
