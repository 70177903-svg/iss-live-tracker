```python
import streamlit as st
import requests
import pandas as pd
import time

# Page Configuration (Browser tab settings)
st.set_page_config(
    page_title="ISS Live Tracking Dashboard",
    page_icon="🚀",
    layout="wide"
)

# App Title & Description
st.title("🚀 International Space Station (ISS) Live Tracker")
st.markdown("This dashboard fetches real-time orbital data from the ISS API and plots its current location live on the map.")

# Creating placeholders for real-time updates
metrics_placeholder = st.empty()
map_placeholder = st.empty()

url = "http://api.open-notify.org/iss-now.json"

# Auto-refresh loop
while True:
    try:
        response = requests.get(url)
        if response.status_code == 200:
            data = response.json()

            # Extract coordinates
            lat = float(data["iss_position"]["latitude"])
            lon = float(data["iss_position"]["longitude"])
            ts = data["timestamp"]

            # 1. Display Metrics side-by-side
            with metrics_placeholder.container():
                col1, col2, col3 = st.columns(3)
                col1.metric(label="🛰️ Latitude", value=f"{lat} °N" if lat > 0 else f"{abs(lat)} °S")
                col2.metric(label="🌐 Longitude", value=f"{lon} °E" if lon > 0 else f"{abs(lon)} °W")
                col3.metric(label="⏰ UNIX Timestamp", value=ts)

            # 2. Display Map (Streamlit requires a DataFrame with 'lat' and 'lon' columns)
            df = pd.DataFrame([[lat, lon]], columns=['lat', 'lon'])
            with map_placeholder.container():
                st.subheader("Current Orbital Position")
                st.map(df, zoom=2)

        else:
            st.error("Error: Unable to fetch data from the ISS API.")

    except Exception as e:
        st.error(f"An error occurred: {e}")

    # Refresh every 5 seconds
    time.sleep(5)
    st.rerun()
```
